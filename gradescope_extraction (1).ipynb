```python
from PIL import Image
import pytesseract
import re
import json

def extract_assignments_and_due_dates_refined_v9(image_path):
    """
    Extracts assignment names and due dates based on the observed
    order: Names, Statuses, Due Dates (refined due date extraction).
    """
    img = Image.open("gradescope_hw.png")
    text = pytesseract.image_to_string(img)
    lines = text.strip().split('\n')

    assignments = []
    name_start_index = -1
    status_start_index = -1
    due_date_start_index = -1

    for i, line in enumerate(lines):
        if "= Name" in line and name_start_index == -1:
            name_start_index = i + 1
        elif "> Status" in line and status_start_index == -1:
            status_start_index = i + 1
        elif "Released Due (CDT)" in line and due_date_start_index == -1:
            due_date_start_index = i + 1
            break

    if name_start_index == -1 or status_start_index == -1 or due_date_start_index == -1:
        print("Could not find the expected headers ('= Name', '> Status', 'Released Due (CDT)').")
        return []

    names = [line.strip() for line in lines[name_start_index:status_start_index] if line.strip() and not re.match(r'^\d+\.?\d* / \d+\.?\d*.*$', line)]
    due_date_lines = lines[due_date_start_index:]
    due_dates = []

    for line in due_date_lines:
        date_match = re.search(r'(Jan|Feb|Mar|Apr|May|Jun|Jul|Aug|Sep|Oct|Nov|Dec)\s+\d{1,2}\s+at\s+\d{1,2}:\d{2}(AM|PM)', line)
        relative_match = re.search(r'(\d+)\s+days?,\s*(\d+)\s+hours?\s+left', line)
        relative_single_match = re.search(r'(\d+)\s+(day|hour)s?\s+left', line)

        if date_match:
            due_dates.append(date_match.group(0).replace(" at", "").split()[-3:])
        elif relative_match:
            days = relative_match.group(1)
            hours = relative_match.group(2)
            if days and hours:
                due_dates.append(f"{days} days, {hours} hours left")
            elif days:
                due_dates.append(f"{days} days left")
            elif hours:
                due_dates.append(f"{hours} hours left")
        elif relative_single_match:
            amount = relative_single_match.group(1)
            unit = relative_single_match.group(2)
            due_dates.append(f"{amount} {unit}s left")
        elif re.search(r'(Sep|May|Apr|Mar)\s+\d{1,2}\s+at\s+\d{1,2}:\d{2}(AM|PM)', line):
            due_dates.append(re.search(r'(Sep|May|Apr|Mar)\s+\d{1,2}\s+at\s+\d{1,2}:\d{2}(AM|PM)', line).group(0).replace(" at", "").split()[-3:])


    # Pair names with due dates (assuming they appear in the same order)
    min_len = min(len(names), len(due_dates))
    for i in range(min_len):
        due_date_str = " ".join(due_dates[i]) if isinstance(due_dates[i], list) else due_dates[i]
        assignments.append({"assignment": names[i], "due_date": due_date_str})

    return assignments

# (rest of the code for create_html_table and save_to_json remains the same)

if __name__ == "__main__":
    image_file = "gradescope_hw.png"
    extracted_data = extract_assignments_and_due_dates_refined_v9(image_file)

    if extracted_data:
        print("Extracted Assignments and Due Dates (Refined v9):")
        for item in extracted_data:
            print(f"Assignment: {item['assignment']}, Due Date: {item['due_date']}")

        html_table = create_html_table(extracted_data)
        with open("assignments.html", "w") as f:
            f.write(html_table)
        print("\nHTML table saved to assignments.html")

        save_to_json(extracted_data)
        print("JSON data saved to assignments.json")
    else:
        print("Could not extract assignment and due date information.")
```
